```python
from ollama import embeddings
from numpy import array, dot
from numpy.linalg import norm
from ollama import chat


EMBEDDING_MODEL = "qwen3-embedding:0.6b"
LANGUAGE_MODEL = "llama3.1:8B"


def cosine_similarity(a, b):
    return round( dot(a, b) / (norm(a) * norm(b)), 2)


prompt = "Im Herbst fallen die Blätter von den Bäumen."
text_2 = "Im Herbst fallen immer wieder Blätter zu Boden, aber dabei explodieren sie in einer wahnsinnigen Wasserfontäne."
text_3 = "Beim Rennen quietschen die Reifen der Autos."
```

```python
response_p = embeddings(model=EMBEDDING_MODEL, prompt=prompt)
response_2 = embeddings(model=EMBEDDING_MODEL, prompt=text_2)
response_3 = embeddings(model=EMBEDDING_MODEL, prompt=text_3)

emb_p = response_p.embedding
emb_2 = response_2.embedding
emb_3 = response_3.embedding
```

```python
cosine_similarity(emb_p, emb_2)
```

```python
cosine_similarity(emb_p, emb_3)
```

```python
print (
    "Wie ähnlich sind Prompt und Dokument 2?",
    cosine_similarity(emb_p, emb_2)
)


print (
    "Wie ähnlich sind Prompt und Dokument 3?",
    cosine_similarity(emb_p, emb_3)
)
```

```python
rag_prompt = f"Du bist ein hilfreicher Assistent. Du bekommst von deinem User folgende Frage: {prompt}. Bearbeite diesen Prompt unter Zuhilfenahme folgender Dokumente: {text_2}"
```

```python
chat_response = chat(
    model = LANGUAGE_MODEL,
    messages = [
        {"role": "user", "content": rag_prompt}
    ]
)
```
